```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Count employees who left and stayed
print(df["Attrition"].value_counts())

# Calculate attrition percentage
print("\nAttrition Percentage:")
print((df["Attrition"].value_counts(normalize=True) * 100).round(2))
```

```python
# Check missing values in each column
print("Missing Values:")
print(df.isnull().sum().sum())
```

```python
# Check duplicate rows

print("Duplicate Rows:")
print(df.duplicated().sum())
```

```python

# 1. Check missing values
print("Total Missing Values:", df.isnull().sum().sum())

# 2. Check duplicate rows
print("Total Duplicate Rows:", df.duplicated().sum())

# 3. Check dataset dimensions
print("Dataset Shape:", df.shape)
```

```python
# Count employees in each attrition category
attrition_counts = df["Attrition"].value_counts()
```

```python
# Create a bar chart
plt.figure(figsize=(7, 5))
sns.barplot(x=attrition_counts.index, y=attrition_counts.values)

plt.title("Employee Attrition Count")
plt.xlabel("Attrition")
plt.ylabel("Number of Employees")

plt.show()
```

```python

# Count employees by department and attrition
department_attrition = pd.crosstab(
    df["Department"],
    df["Attrition"]
)

print(department_attrition)
```

```python

department_attrition.plot(
    kind="bar",
    figsize=(9, 5)
)

plt.title("Employee Attrition by Department")
plt.xlabel("Department")
plt.ylabel("Number of Employees")
plt.xticks(rotation=0)
plt.legend(title="Attrition")
plt.tight_layout()
plt.show()
```

```python

# Calculate attrition rate for each department
department_rate = (
    df.groupby("Department")["Attrition"]
    .apply(lambda x: (x == "Yes").mean() * 100)
    .round(2)
)

print("Department-wise Attrition Rate (%):")
print(department_rate)
```

```python

department_rate.plot(
    kind="bar",
    figsize=(8, 5)
)

plt.title("Department-wise Attrition Rate")
plt.xlabel("Department")
plt.ylabel("Attrition Rate (%)")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python

# Calculate attrition rate based on overtime
overtime_rate = (
    df.groupby("OverTime")["Attrition"]
    .apply(lambda x: (x == "Yes").mean() * 100)
    .round(2)
)

print("Attrition Rate by Overtime (%):")
print(overtime_rate)
```

```python

overtime_rate.plot(kind="bar", figsize=(7, 5))

plt.title("Overtime vs Employee Attrition Rate")
plt.xlabel("Overtime")
plt.ylabel("Attrition Rate (%)")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python

# Calculate attrition rate by job satisfaction
satisfaction_rate = (
    df.groupby("JobSatisfaction")["Attrition"]
    .apply(lambda x: (x == "Yes").mean() * 100)
    .round(2)
)

print("Attrition Rate by Job Satisfaction (%):")
print(satisfaction_rate)
```

```python

satisfaction_rate.plot(
    kind="bar",
    figsize=(8, 5)
)

plt.title("Job Satisfaction vs Employee Attrition")
plt.xlabel("Job Satisfaction Level")
plt.ylabel("Attrition Rate (%)")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python

income_analysis = df.groupby("Attrition")["MonthlyIncome"].mean().round(2)

print("Average Monthly Income by Attrition:")
print(income_analysis)
```

```python

income_analysis.plot(kind="bar", figsize=(7, 5))

plt.title("Average Monthly Income vs Employee Attrition")
plt.xlabel("Attrition")
plt.ylabel("Average Monthly Income")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python

# Create age groups
df["AgeGroup"] = pd.cut(
    df["Age"],
    bins=[17, 25, 35, 45, 55, 65],
    labels=["18-25", "26-35", "36-45", "46-55", "56-65"]
)

# Calculate attrition rate for each age group
age_attrition = (
    df.groupby("AgeGroup", observed=False)["Attrition"]
    .apply(lambda x: (x == "Yes").mean() * 100)
    .round(2)
)

print("Age-wise Attrition Rate (%):")
print(age_attrition)
```

```python

age_attrition.plot(kind="bar", figsize=(8, 5))

plt.title("Age Group vs Employee Attrition Rate")
plt.xlabel("Age Group")
plt.ylabel("Attrition Rate (%)")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python

# Overall attrition rate
overall_rate = (df["Attrition"] == "Yes").mean() * 100

# Department with highest attrition rate
highest_department = department_rate.idxmax()
highest_department_rate = department_rate.max()

# Overtime group with higher attrition rate
higher_overtime_group = overtime_rate.idxmax()
higher_overtime_rate = overtime_rate.max()

# Job satisfaction level with highest attrition rate
highest_satisfaction = satisfaction_rate.idxmax()
highest_satisfaction_rate = satisfaction_rate.max()

print(f"Overall Attrition Rate: {overall_rate:.2f}%")
print(f"Highest Department Rate: {highest_department} ({highest_department_rate:.2f}%)")
print(f"Higher Attrition Overtime Group: {higher_overtime_group} ({higher_overtime_rate:.2f}%)")
print(f"Highest Attrition Satisfaction Level: {highest_satisfaction} ({highest_satisfaction_rate:.2f}%)")
```
